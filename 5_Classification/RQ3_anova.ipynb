```python
df_binary = df_result[df_result["Number of Classes"] == 2]
df_binary
```

```python
def oneway_anova(df_result, cv_method, target_column):
    from scipy.stats import f_oneway

    df_anova= pd.DataFrame(columns=['CV','F-statistic', 'P-value', 'Significance'])
    for method in cv_method:
        print(f"Cv : {method}")
        df_tmp = df_result[df_result['CV'] == method] 
        value = df_tmp[target_column].unique()
        count = df_tmp[target_column].nunique()
        if count == 2:
            # Perform ANOVA
            result = f_oneway(df_tmp[df_tmp[target_column] == value[0]]['Accuracy'], 
                            df_tmp[df_tmp[target_column] == value[1]]['Accuracy'])
        elif count == 3:
            result = f_oneway(df_tmp[df_tmp[target_column] == value[0]]['Accuracy'], 
                            df_tmp[df_tmp[target_column] == value[1]]['Accuracy'],
                             df_tmp[df_tmp[target_column] == value[2]]['Accuracy'])
            
        # Extracting F-statistic and p-value
        f_statistic = result.statistic
        p_value = result.pvalue

        print("F-statistic:", f_statistic)
        print("P-value:", p_value)

        if p_value < 0.05:
            p_value_result = "Yes"
        else:
            p_value_result = "No"

        df_anova.loc[len(df_anova)]= [method, f_statistic.round(2), p_value.round(2), p_value_result]
    return df_anova
```
